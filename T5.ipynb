# Python Lists



## Creating, Indexing & Slicing

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
print(numbers[::-1])
```

## Adding & Removing

```python
numbers = [1, 2, 3]
numbers.append(4)
numbers.extend([5, 6])
numbers.insert(1, 10)
print(numbers)

numbers.remove(10)
x = numbers.pop()
print(numbers)
print("Removed:", x)
```

### `append()` vs `extend()`

```python
a = [1, 2, 3]
a.append([4, 5])
print(a)

b = [1, 2, 3]
b.extend([4, 5])
print(b)
```

## Useful Methods

```python
numbers = [50, 10, 40, 10, 20]
print(len(numbers))
print(numbers.count(10))
print(numbers.index(40))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
```

## `sort()` vs `sorted()`

```python
numbers = [50, 10, 40, 20, 30]
result = numbers.sort()
print(numbers)
print(result)  # None

numbers = [50, 10, 40, 20, 30]
result = sorted(numbers)
print(numbers)  # original unchanged
print(result)
```

## Copying Lists — Important

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
print(b)
```

```python
a = [1, 2, 3]
b = a.copy()
b.append(4)
print(a)
print(b)
```

## Nested Lists

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1])
print(matrix[1][2])

matrix[0][1] = 100
print(matrix)
```

## List Comprehension

```python
numbers = [1, 2, 3, 4, 5]
squares = [x * x for x in numbers]
even = [x for x in numbers if x % 2 == 0]
print(squares)
print(even)
```

## `map()` and `filter()`

```python
numbers = [1, 2, 3, 4, 5]
doubled = map(lambda x: x * 2, numbers)
even = filter(lambda x: x % 2 == 0, numbers)
print(type(doubled))
print(list(doubled))
print(list(even))
```

## Coding Challenges

1. Find the largest element **without using `max()`**.
2. Remove duplicates while preserving order. Example: `[3,1,3,2,1] → [3,1,2]`
3. Find the **second-largest distinct** element without `sorted()`, `max()`, or `min()`.
4. Rotate a list right by `k`. Example: `[1,2,3,4,5], k=2 → [4,5,1,2,3]`
5. Find pairs whose sum equals a target. Example: `[2,7,11,15,3,6]`, target `9` → `(2,7), (3,6)`

## Quick Revision

- `append(x)` → adds **one element**
- `extend(iterable)` → adds **multiple elements**
- `remove(x)` → removes by value
- `pop()` → removes and returns an element
- `sort()` → modifies the list, returns `None`
- `sorted()` → returns a new sorted list
- `copy()` → creates a shallow copy
- Lists are **mutable**
- `map()` → returns a `map` iterator
- `filter()` → returns a `filter` iterator
