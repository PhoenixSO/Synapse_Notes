Q1. Body Mass Index (BMI) is a simple measure that helps determine whether a person's weight is appropriate for their height.

Write a Python program that takes the BMI value as input from the user and displays the corresponding weight category using the following rules:

BMI Value       Category
Less than 18.5  Underweight
18.5 to 24.9    Normal Weight
25.0 to 29.9    Overweight
30.0 or above   Obese

```python
bmi = float(input())
if bmi < 18.5:
    print("Underweight")
elif bmi <= 24.9:
    print("Normal Weight")
elif bmi <= 29.9:
    print("Overweight")
else:
    print("Obese")
```
