# Elif Keyword
### The elif keyword allows you to check multiple expressions for True and execute a block of code as soon as one of the conditions evaluates to True.

```python
a = 33
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
```

## Multiple Elif Statements

```python
score = 75

if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
elif score >= 60:
  print("Grade: D")
```

## Else Keyword
### The else keyword catches anything which isn't caught by the preceding conditions.

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

## Else Without Elif

```python
a = 200
b = 33
if b > a:
  print("b is greater than a")
else:
  print("b is not greater than a")
```

### Quiz
**Q20. In the marks/attendance example above, using the same `if attendance < 75 ... elif marks >= 90 ... elif marks >= 75 ... else` chain, suppose `attendance = 60` and `marks = 95`. What gets printed?**
- A) "Grade A", because marks is very high
- B) "Not eligible due to low attendance"
- C) Both "Grade A" and "Not eligible due to low attendance" print
- D) "Fail", because attendance is too low

<details><summary>Answer</summary>B) "Not eligible due to low attendance"

*In an if/elif chain, Python checks conditions top to bottom and stops at the FIRST one that's True. Since attendance < 75 is True, none of the marks conditions are even looked at — high marks don't matter here.*</details>
