# Python Functions

#### Functions let us package a sequence of instructions into a reusable block of code.

A function can **take inputs**, perform some operation, and optionally **return a result**.

Basic syntax:

```python
def function_name(parameters):
    # function body
    return result
```

## Defining and Calling a Function

````python
def greet():
    print("Hello, Python!")

greet()
greet()
````

A function is not executed when it is defined. Its body runs when the function is **called**.

### Quiz

```python

    print("A")

print("B")
test()
print("C")
```

- A) A B C
- B) B A C
- C) B C A
- D) A C B

<details><summary>Answer</summary>B) B A C

*The function definition does not print anything. `print("B")` runs first, then `test()` prints `A`, then `C`.*</details>

## Parameters and Arguments

````python
def introduce(name, age):
    print("Name:", name)
    print("Age:", age)

introduce("Riya", 19)
````

Here, `name` and `age` are **parameters** of the function.

`"Riya"` and `19` are **arguments** supplied when the function is called.

### Quiz

**Which statement is correct?**

- A) A parameter is the value passed during a call
- B) An argument is a variable written in the function definition
- C) A parameter receives an argument when the function is called
- D) Parameters and arguments are completely unrelated

<details><summary>Answer</summary>C) A parameter receives an argument when the function is called

*Parameters are the function's named inputs; arguments are the actual values supplied to those inputs.*</details>

## Return Statement

````python
def square(n):
    return n * n

result = square(7)
print(result)

# A returned value can also be used directly
print(square(4) + square(3))
````

`return` sends a value back to the place where the function was called.

This is different from `print()`: `print()` displays a value, while `return` gives a value back to the calling code.

## Return Ends the Function

````python
def check_number(n):
    if n < 0:
        return "negative"

    print("This line runs for non-negative numbers.")
    return "non-negative"

#print(check_number(-3))
print(check_number(5))
````

### Quiz

**What happens when `check_number(-3)` is called?**

- A) It prints both messages
- B) It immediately returns `"negative"`
- C) It returns `"non-negative"`
- D) It produces a SyntaxError

<details><summary>Answer</summary>B) It immediately returns `"negative"`

*Executing `return` exits the function immediately, so the later `print()` is skipped.*</details>

## Multiple Return Values

````python
def calculate(a, b):
    return a + b, a - b, a * b

total, difference, product = calculate(10, 3)

print(total)
print(difference)
print(product)
````

### Quiz

**What is actually returned by `return a + b, a - b`?**

- A) Two separate return statements
- B) A list
- C) A tuple
- D) A dictionary

<details><summary>Answer</summary>C) A tuple

*Python packs comma-separated returned values into a tuple, which can then be unpacked into multiple variables.*</details>

## Positional and Keyword Arguments

````python
def student(name, age, branch):
    print(name, age, branch)

student("Ananya", 19, "CSE")

student(name="Ananya", branch="CSE", age=19)
````

Keyword arguments let you specify which parameter receives a value, so their order can be changed.

- C) 1 3 2

<details><summary>Answer</summary>B) 1 2 3

*The keywords explicitly map `a=1`, `b=2`, and `c=3`.*</details>

## Default Arguments

````python
def power(number, exponent=2):
    return number ** exponent

print(power(5))
print(power(5, 3))
print(power(number=2, exponent=5))
````

### Quiz

**What does `power(4)` return when the function is defined as `def power(number, exponent=2)`?**

- A) 2
- B) 4
- C) 8
- D) 16

<details><summary>Answer</summary>D) 16

*Because no exponent is supplied, the default value `2` is used: `4 ** 2 = 16`.*</details>

## Interesting Question: Default Values Are Evaluated Once

````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("A"))
print(add_item("B"))
print(add_item("C"))
````

This is a classic Python pitfall: the default list is created **once**, when the function is defined, rather than being recreated for every call.

### Quiz

**What is printed by the code above?**

- A) ['A'] / ['B'] / ['C']
- B) ['A'] / ['A', 'B'] / ['A', 'B', 'C']
- C) ['A', 'B', 'C'] / ['A', 'B', 'C'] / ['A', 'B', 'C']
- D) An error

<details><summary>Answer</summary>B) ['A'] / ['A', 'B'] / ['A', 'B', 'C']

*All three calls share the same default list. A safer pattern is `items=None`, followed by `items = []` inside the function.*</details>

## `*args`: Variable Number of Positional Arguments

````python
def total(*numbers):
    print("Received:", numbers)
    return sum(numbers)

print(total(2, 4))
print(total(1, 2, 3, 4, 5))
````

`*args` collects any extra positional arguments into a tuple.

### Quiz

**Inside `def show(*values):`, what type is `values`?**

- A) list
- B) tuple
- C) set
- D) dictionary

<details><summary>Answer</summary>B) tuple

*`*args` collects positional arguments into a tuple.*</details>

## `**kwargs`: Variable Number of Keyword Arguments

````python
def profile(**details):
    for key, value in details.items():
        print(key, "=", value)

profile(name="Aarav", branch="CSE", year=2)
````

`**kwargs` collects keyword arguments into a dictionary.

### Quiz

**What is the type of `details` inside `def profile(**details)`?**

- A) tuple
- B) list
- C) dictionary
- D) set

<details><summary>Answer</summary>C) dictionary

*Keyword arguments are collected as key-value pairs in a dictionary.*</details>

## Combining Parameters, `*args`, and `**kwargs`

````python
def report(name, *marks, **info):
    print("Name:", name)
    print("Marks:", marks)
    print("Extra info:", info)

report("Meera", 91, 88, 95, branch="CSE", year=2)
````

### Quiz

**In the call above, what is stored in `marks`?**

- A) ('Meera',)
- B) (91, 88, 95)
- C) [91, 88, 95]
- D) {'branch': 'CSE', 'year': 2}

<details><summary>Answer</summary>B) (91, 88, 95)

*`name` receives the first positional argument, `*marks` collects the remaining positional arguments, and `**info` collects the keyword arguments.*</details>

## Scope: Local vs Global Variables

````python
x = 10

def demo():
    x = 20
    print("Inside:", x)

demo()
print("Outside:", x)
````

### Quiz

**What is printed?**

- A) Inside: 20 and Outside: 20
- B) Inside: 10 and Outside: 10
- C) Inside: 20 and Outside: 10
- D) A NameError occurs

<details><summary>Answer</summary>C) Inside: 20 and Outside: 10

*The `x` inside the function is a local variable and does not replace the global `x`.*</details>

## The `global` Keyword

````python
score = 10

def increase_score():
    global score
    score += 5

increase_score()
print(score)
````

### Quiz

**What is printed?**

- A) 10
- B) 15
- C) 5
- D) UnboundLocalError

<details><summary>Answer</summary>B) 15

*`global score` tells Python that the function should modify the global variable rather than create a local one.*</details>

## Nested Functions

````python
def outer():
    message = "Hello"

    def inner():
        print(message)

    inner()

outer()
````

An inner function can access variables from its enclosing function. This is part of Python's lexical scope rules.

### Quiz

**What does `inner()` access when it prints `message`?**

- A) Only global variables
- B) The variable from `outer()`
- C) Only variables passed as arguments
- D) It cannot access `message`

<details><summary>Answer</summary>B) The variable from `outer()`

*The nested function can read variables from its enclosing scope.*</details>

## Lambda Functions

````python
square = lambda x: x * x

print(square(6))

add = lambda a, b: a + b
print(add(3, 7))
````

A lambda is a small anonymous function containing a single expression.

### Quiz

- D) def f(x, y): return x * y

<details><summary>Answer</summary>B) def f(x): return x * 2

*The lambda takes one argument and returns `x * 2`.*</details>

## Functions as Arguments

````python
def apply_twice(function, value):
    return function(function(value))

def increment(x):
    return x + 1

print(apply_twice(increment, 5))
````

### Quiz

**What is printed?**

- A) 6
- B) 7
- C) 10
- D) 25

<details><summary>Answer</summary>B) 7

*`increment(5)` gives 6, and then `increment(6)` gives 7.*</details>

## `map()` with a Function

````python
numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))

print(squares)
````

### Quiz

**What is the value of `squares`?**

- B) [2, 4, 6, 8, 10]
- C) [1, 4, 9, 16, 25]
- D) 25

<details><summary>Answer</summary>C) [1, 4, 9, 16, 25]

*`map()` applies the lambda to every element.*</details>

## Recursion

````python
def countdown(n):
    if n == 0:
        print("Go!")
        return

    print(n)
    countdown(n - 1)

countdown(3)
````

A recursive function calls itself.

A recursive solution needs a **base case** so that the calls eventually stop.

### Quiz

**What is the output of `countdown(3)`?**

- A) 3 2 1 Go!
- B) 3 2 1 0 Go!
- C) Go! 1 2 3
- D) 3 2 1

<details><summary>Answer</summary>A) 3 2 1 Go!

*The base case is reached when `n == 0`; it prints `Go!` and returns.*</details>

## Interesting Recursion Question: Factorial

````python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

### Quiz

**Which expression is evaluated first when calculating `factorial(4)`?**

- A) 4 * 3
- B) factorial(3)
- C) factorial(0)
- D) 1 * 2 * 3 * 4

<details><summary>Answer</summary>B) factorial(3)

*The function first evaluates the recursive call `factorial(3)` before multiplying its result by 4.*</details>

## Guess the Output — Challenge Round

````python
def mystery(a, b=5):
    a += b
    return a

x = 10
y = mystery(x)
print(x, y)
````

### Quiz

**What is printed?**

- A) 10 15
- B) 15 15
- C) 10 10
- D) 15 10

<details><summary>Answer</summary>A) 10 15

*Integers are immutable. The function computes `a + b` and returns 15, but the original `x` remains 10.*</details>

````python
def change(data):
    data.append(100)

numbers = [1, 2, 3]
change(numbers)

print(numbers)
````

````python
def f(x):
    if x > 0:
        return x
    return -x

print(f(-8) + f(3))
````

## Quick Revision

| Concept | Key idea |
|---|---|
| `def` | Defines a function |
| Function call | Executes the function |
| Parameter | Variable in the function definition |
| Argument | Actual value passed to the function |
| `return` | Sends a value back to the caller |
| Default argument | Used when an argument is omitted |
| Keyword argument | Argument supplied using its parameter name |
| `*args` | Collects extra positional arguments into a tuple |
| `**kwargs` | Collects extra keyword arguments into a dictionary |
| Local scope | Variable available inside its function |
| `global` | Allows a function to rebind a global variable |
| Lambda | Small anonymous function |
| Recursion | Function calling itself |
| Higher-order function | Function that accepts/returns another function |
